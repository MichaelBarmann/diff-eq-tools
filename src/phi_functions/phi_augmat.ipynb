### `phi_augmat.ipynb`

This notebook implements functions to compute $\phi_k(B)$ for a general square matrix $B$ using the *augmented matrix exponential* approach: <br/>

----------------------------------------------------------------------------------------------------------------------
**The Augmented Matrix Exponential:** <br/>
Let $C_k$ be the $N(k+1) \times N(k+1)$ block diagonal matrix $C_k := \textrm{diag}(B,I^{(1)},...,I^{(k)})$, where the $I^{(j)}$'s are $N \times N$ identity blocks. Then the first $N$ rows of the matrix $e^C$ is exactly the block matrix $[\phi_0(B) \; \phi_1(B) \; \cdots \; \phi_k(B)]$.  

<div><strong>Pros:</strong></div>
<ul style="margin-top: 0;">
  <li>Works for <em>any</em> square matrix $B$; does <em>not</em> need to be diagonal or even invertible!</li>
  <li>Numerically stable!</li>
</ul>

<div><strong>Cons:</strong></div>
<ul style="margin-top: 0;">
  <li>Becomes expensive as the size of $B$ gets large.</li>
</ul>

```julia
using LinearAlgebra, Random, Statistics, NBInclude
```

```julia
function phis_augmat(B::AbstractMatrix, k::Int64)
    """
    Computes φ_0(B),...,φ_k(B) for an arbitrary square matrix B (with real or complex entries) and a nonnegative integer k. 
    
    PARAMETERS
    ----------
    B :: An N x N matrix of real or complex entries
    k :: A nonnegative integer
    
    RETURNS
    --------
    """
    #Error handling
    size(B,1) == size(B,2) || throw(ArgumentError("B must be a square matrix. Passed B of size $(size(B))."))
    0 ≤ k ≤ 3 || throw(ArgumentError("k must be a nonnegative integer between 0 and 3."))

    #Matrix type conversion (to be safe) 
    T = promote_type(eltype(B), Float64)
    B = Matrix{T}(B)      

    #Create block matrix 
    N = size(B,1) 
    C = zeros(T, N*(k+1), N*(k+1))
    C[1:N, 1:N] .= B 

    Iₙ = Matrix{T}(I, N, N)

    for j=1:k
        C[(j-1)*N+1:j*N, j*N+1:(j+1)*N] .= Iₙ
    end

    E = exp(C) 
    phi_mats = [E[1:N, j*N+1:(j+1)*N] for j=0:k]
    return phi_mats
end 
```

```julia
#Convenience function that returns the single matrix φₖ(B)
phi_augmat(B::AbstractMatrix, k::Int) = phis_augmat(B, k)[k+1]
```
