### `test_problems.ipynb`
*Created: Sept 22, 2026* <br/>
This notebook contains a bunch of semilinear ODEs of the form 
\begin{align*}
   u_t = Au + f(u,t)
\end{align*}
where $A$ is an $N \times N$ matrix and $f:\mathbb{R}^{N} \times [0,\infty) \to \mathbb{R}$ is a (possibly nonlinear) function. These ODEs are for testing the accuracy and order of convergence of some ETD solvers.
