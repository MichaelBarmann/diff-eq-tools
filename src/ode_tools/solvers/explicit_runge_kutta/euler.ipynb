In this notebook we define the ExplicitEuler type, and we provide an implementation of first-order Euler's method (which we note is technically a first-order Runge-Kutta method). 

````julia
using NBInclude
@nbinclude("runge_kutta.ipynb")
````

````julia
function euler(f::F, u0::U, tspan::NTuple{2,Float64}, p = nothing; dt::Float64) where {F, U}

    """
    Solves the IVP du/dt = f(u, p, t), u(t0) = u0 on the interval tspan, where t0 = tspan[1]. 

    PARAMETERS
    ----------
    f:  The right-hand side of the ODE. Must have two arguments, the first being time. 
    u0: initial condition 
    tspan :: time interval over which to integrate
    dt :: the fixed time step

    RETURNS
    -------
    sol :: the ODE solution as an array of scalars (if f is scalar-valued) 
           or an array of arrays (if f is vector-valued)
    """

    t0, tf = tspan 

    #Number of steps/iterations
    nsteps = floor(Int, (tf - t0) / dt)

    #Time values at which the solution will be recorded
    t = collect(range(t0, t0 + nsteps * dt, length = nsteps + 1))

    #NOTE: t[end] = t0 + nsteps * dt ≤ tf (with equality iff (tf - t0) / dt is an integer)
     
    #Array to store the numerical solution 
    u = [zero(u0) for _ in 1:nsteps + 1]
    u[1] = u0 
     
    #Euler's method 
    for n=1:nsteps
        u[n+1] = u[n] + dt*f(u[n], p, t[n])
    end 
    
    return (u = u, t = t)
    
end
````
